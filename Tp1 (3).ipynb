```python
text = "Bonjour ! Le Traitement Automatique du Langage Naturel (NLP) est VRAIMENT passionnant en 2026, n'est-ce pas ?"

text_lower = text.lower()

text_clean = text_lower.translate(str.maketrans('', '', string.punctuation))

print("Texte original  :", text)
print("Texte nettoyé   :", text_clean)
```

```python

tokens = word_tokenize(text_clean, language='french')
print("Tokens originaux")
print(tokens)

stop_words = set(stopwords.words('french'))

stop_words.add("vraiment")

filtered_tokens = [w for w in tokens if w not in stop_words]

print("\n Tokens après suppression des stopwords")
print(filtered_tokens)

nb_originaux = len(tokens)
nb_filtres = len(filtered_tokens)
nb_supprimes = nb_originaux - nb_filtres
proportion = (nb_supprimes / nb_originaux) * 100

print(f"\nNombre de tokens au début : {nb_originaux}")
print(f"Nombre de tokens supprimés : {nb_supprimes}")
print(f"Proportion supprimée : {proportion:.2f}%")
```

```python
nlp = spacy.load('fr_core_news_sm')

text_filtered = " ".join(filtered_tokens)
doc = nlp(text_filtered)

print(f"{'Mot filtré (Avant)':<20} ---> {'Lemme (Après)'}")
print("-" * 45)

lemmatized_tokens = []
for token in doc:
    lemmatized_tokens.append(token.lemma_)
    print(f"{token.text:<20} ---> {token.lemma_}")
```
